# Pretrained GloVe Embeddings, Similarity, and Word Analogies

This notebook studies what pretrained word embeddings contain before they are used inside a classifier.

The progression is:

```text
pretrained GloVe vectors
→ embedding lookup
→ cosine similarity
→ nearest words
→ vector relationships
→ word analogies
```

Read embeddings to a dictionary

````python
import numpy as np

def read_embeddings(n=1000):
    # Reads n embeddings from file
    # Returns a dictionary where embedding[w] is the embeding of string w
    embedding = {}
    count = 0
    with open('glove.6B.100d.txt', encoding="utf8") as f:
        for line in f:
            count+=1
            ls = line.split(" ")
            emb = np.array([np.float32(x) for x in ls[1:]])
            embedding[ls[0]]=emb
            if count>= n:
                break
    return embedding

vocabulary_size = 200000
embedding = read_embeddings(vocabulary_size)
````

Define similarity metric

````python
def cosine_similarity(w1,w2,emb):
  e1,e2 = emb[w1], emb[w2]
  return np.dot(e1,e2)/np.linalg.norm(e1)/np.linalg.norm(e2)
````

Embeddings even capture the similarity among universities!

````python
words = ['utep','princeton','harvard','nmsu']
for w1 in words:
  for w2 in words:
    print('cosine_similarity({},{})={:4.3f}'.format(w1,w2,cosine_similarity(w1,w2,embedding)))
````

## Nearest Words in the Embedding Space

Write a function that receives a word `w` and an embedding dictionary and returns the word whose embedding is most similar to `w`.

Two versions are compared:

- Euclidean distance
- cosine similarity


````python
def most_similar(w,emb):
  # Use Euclidean distance
  msw, min_dist = '', np.inf
  for w1 in emb:
    if w == w1: continue
    dist = np.linalg.norm(emb[w]-emb[w1])
    if dist < min_dist:
      msw, min_dist = w1, dist
  return msw
````

````python
def most_similar_cs(w,emb):
  # Use cosine similarity
  msw, max_sim = '', -np.inf
  for w1 in emb:
    if w == w1: continue
    sim = cosine_similarity(w,w1,emb)
    if sim>max_sim:
      msw, max_sim = w1, sim
  return msw
````

Using Euclidean Distance

````python
for w in ['platypus','python','federer','taco','black','ford','dog','java','spain']:
  print(f'The most similar word to {w} is {most_similar(w,embedding)}')
````

Using Cosine Similarity

````python
for w in ['platypus','python','federer','taco','black','ford','dog','java','spain']:
  print(f'The most similar word to {w} is {most_similar_cs(w,embedding)}')
````

## Word Analogies with Embedding Arithmetic

Word embeddings can also be used to solve analogy problems such as:

**France is to Paris as Spain is to ?**

Possible answers:

- Washington
- Berlin
- Moscow
- Madrid

The expected answer is Madrid.


### Vector-Offset Idea

The vector connecting related word embeddings can represent an approximate relationship direction.

For the geographic example, the source idea is:

```text
E['france'] - E['paris'] ≈ E['spain'] - E['madrid']
```

which can be rearranged as:

```text
E['madrid'] ≈ E['paris'] - E['france'] + E['spain']
```

Therefore, the vector `E['paris'] - E['france'] + E['spain']` is expected to be closer to `E['madrid']` than to alternatives such as Washington, Berlin, or Moscow.

This vector-offset idea is the basis of the analogy experiments below.


````python
def word_analogy(pair,w,options,emb):
  pair_vector = emb[pair[0]] - emb[pair[1]]
  dests = [emb[op]+pair_vector for op in options]
  dists = [np.linalg.norm(emb[w]-d) for d in dests]
  #print(dists)
  return options[np.argmin(dists)]
````

## What Pretrained Embeddings Learn

GloVe vectors are learned from word co-occurrence statistics in a large corpus. Words that appear in related linguistic contexts can obtain related vectors.

That structure can be explored in several ways:

- cosine similarity measures vector direction similarity,
- Euclidean distance measures geometric distance,
- nearest-word search finds close vectors,
- analogy arithmetic tests whether relationships correspond to useful vector offsets.

Pretrained embeddings therefore provide semantic information before a task-specific classifier is trained.


## Key Takeaways

- Each word is represented by a dense numerical vector rather than a sparse one-hot feature.
- Similarity in embedding space can reflect semantic or contextual similarity.
- Cosine similarity is especially useful for comparing vector direction.
- Word analogies demonstrate that some relationships are approximately encoded as vector directions.
- These pretrained vectors can later be used to initialize embedding layers for text-classification models.
